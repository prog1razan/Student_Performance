## Create a New Column

Create a new column called `average_score`.

The average score should be calculated from:

- `math_score`
- `english_score`
- `science_score`


```python
students["average_score"] = (students["math_score"] + students["english_score"] + students["science_score"]) / 3

students
```

## Add a Pass / Fail Column

Create a new column called `status`.

If the student's `average_score` is **70 or more**, the status should be `"Pass"`.

Otherwise, the status should be `"Fail"`.


```python
students["status"] = "Pass"
students["status"] = students["status"].where(
    students["average_score"] >= 70,
    "Fail"
)

students[["student_name", "average_score", "status"]]
```

## Challenge

Create a column called `performance_level`.

Rules:

- If `average_score` is 90 or more → `"Excellent"`
- If `average_score` is 80 to 89 → `"Very Good"`
- If `average_score` is 70 to 79 → `"Good"`
- Otherwise → `"Needs Improvement"`

Then display:

1. Student name
2. Average score
3. Performance level


```python

students["performance_level"] = "Needs Improvement"

students.loc[students["average_score"] >= 90, "performance_level"] = "Excellent"
students.loc[(students["average_score"] >= 80) & (students["average_score"] < 90), "performance_level"] = "Very Good"
students.loc[(students["average_score"] >= 70) & (students["average_score"] < 80), "performance_level"] = "Good"

students[["student_name", "average_score", "performance_level"]]
```
